```python
def right_angled_triangle(a,b,c):
    sides=[a,b,c]
    sides.sort()
    if sides[0]**2+sides[1]**2==sides[2]**2:
        print(sides, "form the sides of a right angled triangle")
    else:
        print("the given sides do not form a right angled triangle")
right_angled_triangle(13,5,12)
    
    
```
